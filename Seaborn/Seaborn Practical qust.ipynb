```python
import seaborn as sns
import matplotlib.pyplot as plt
```

**Swarm Plot with Hue**

1)Use the penguins dataset to create a swarm plot for body mass (body_mass_g) across different
species (species).



Add hue for gender (sex) to visualize how gender affects body mass within species.

```python
df=sns.load_dataset("penguins")
df
sns.swarmplot(x="species",y="body_mass_g",hue="sex",data=df)
plt.show()
```

**Overlaying Swarm Plot on Boxplot**

Load the iris dataset and create a boxplot to compare sepal length (sepal_length) across species
(species).


Overlay a swarm plot on the boxplot to highlight individual data points.

```python
df1=sns.load_dataset("iris")
df1
sns.boxplot(x="species",y="sepal_length",data=df1)
sns.swarmplot(x="species",y="sepal_length",data=df1)
plt.show()
```

**Categorical Data and Numerical Data**

Using the titanic dataset, create a swarm plot for passenger age (age) by passenger class (class).



Add hue="sex" to differentiate between males and females.

```python
df2=sns.load_dataset("titanic")
df2
sns.swarmplot(x="class",y="age",hue="sex",data=df2)
plt.show()
```

**Combining Swarm Plot with Violin Plot**

Load the diamonds dataset and create a violin plot for diamond price (price) across cut quality (cut).

Overlay a swarm plot for detailed visualization of individual points.

```python
df3=sns.load_dataset("diamonds")
df3
sns.violinplot(x="cut",y="price",data=df3)
sns.swarmplot(x="cut",y="price",data=df3)
plt.show()
```
