#numpy array copy and view

1. Create a NumPy array and make a copy of it. Modify the original array and print both
the original and copied arrays.

```python
import numpy as np
a=np.array([1,2,3,4])
b=a.copy()
print(a)
b[2]=20
print(b)
```

2. Create a view of a NumPy array. Modify the view and print both the original and the
view. How do the results compare to using a copy?


```python
import numpy as np
a=np.array([1,2,3,4])
b=a.view()
b[2]=6
print(a)
print(b)
```

Explain the difference between array.copy() and array.view() in NumPy. Provide an
example to illustrate the difference.

```python
import numpy as np
a=np.array([1,2,3,4,5])
b=np.array([1,2,3,4,5])
c=a.copy()
a[0]=10
d=b.view()
b[0]=20
print(c.base)
print(d.base)
```
