```python
Q2
def atm_withdrawal(withdrawal_amount):
    account_balance = 5000000

    if withdrawal_amount % 100 != 0:
        return "Error: Withdrawal amount must be in multiples of 100."

    if withdrawal_amount > account_balance:
        return "Error: Insufficient funds."

    new_balance = account_balance - withdrawal_amount

    return f"Withdrawal successful! Your new balance is: {new_balance}"

withdrawal_amount = int(input("Enter the withdrawal amount: "))
result = atm_withdrawal(withdrawal_amount)
print(result)
```

```python
Q5
def calculate_grade(marks):
    if marks >= 90:
        return "Grade A"
    elif marks >= 80:
        return "Grade B"
    elif marks >= 70:
        return "Grade C"
    else:
        return "Grade D"

marks = int(input("Enter the student's marks: "))
grade = calculate_grade(marks)
print(f"The student's grade is: {grade}")
```

```python
#Q6

def count_word_occurrences(sentence, word):

    sentence = sentence.lower()
    word = word.lower()

    words = sentence.split()
    return words.count(word)

sentence = "This is a test. This test is simple."
word = "test"
print(f"The word '{word}' appears {count_word_occurrences(sentence, word)} times.")
```

```python
#Q7
def bmi_calculator(weight,
height):
    if height <= 0:
        return "Height must be greater than 0."

    bmi = weight / (height ** 2)

    if bmi < 18.5:
        category = "Underweight"
    elif 18.5 <= bmi < 25:
        category = "Normal weight"
    elif 25 <= bmi < 30:
        category = "Overweight"
    else:
        category = "Obese"

    return round(bmi, 2), category

weight = 70
height = 1.75
bmi, category = bmi_calculator(weight, height)
print(f"BMI: {bmi}, Category: {category}")
```

```python
#Q8)
def sum_of_digits(number):

    number = abs(number)

    return sum(int(digit) for digit in str(number))

number = 12345
print(f"The sum of digits in {number} is {sum_of_digits(number)}.")
```

```python
#Q9)

def add_expense(expenses):

    name = input("Enter expense name: ")
    category = input("Enter expense category (e.g., Food, Transport, Bills): ")
    try:
        amount = float(input("Enter expense amount: "))
    except ValueError:
        print("Invalid amount. Please enter a number.")
        return

    expenses.append({"name": name, "category": category, "amount": amount})
    print("Expense added successfully!\n")


def view_expenses(expenses):

    if not expenses:
        print("No expenses to show.\n")
        return

    print(f"{'Name':<20} {'Category':<15} {'Amount':<10}")
    print("-" * 45)
    for expense in expenses:
        print(f"{expense['name']:<20} {expense['category']:<15} {expense['amount']:<10.2f}")
    print()


def total_by_category(expenses):

    if not expenses:
        print("No expenses to calculate totals.\n")
        return

    category_totals = {}
    for expense in expenses:
        category_totals[expense["category"]] = category_totals.get(expense["category"], 0) + expense["amount"]

    print(f"{'Category':<15} {'Total Amount':<10}")
    print("-" * 30)
    for category, total in category_totals.items():
        print(f"{category:<15} {total:<10.2f}")
    print()


def main():

    expenses = []

    while True:
        print("Expense Tracker")
        print("1. Add Expense")
        print("2. View Expenses")
        print("3. View Total by Category")
        print("4. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_expense(expenses)
        elif choice == "2":
            view_expenses(expenses)
        elif choice == "3":
            total_by_category(expenses)
        elif choice == "4":
            print("Exiting Expense Tracker. Goodbye!")
            break
        else:
            print("Invalid choice. Please try again.\n")


if __name__ == "__main__":
    main()
```

```python
#Q10)
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author, year):
        self.books.append({"title": title, "author": author, "year": year, "borrowed": False})
        print(f"Book '{title}' added successfully!")

    def search_book(self, title):
        found_books = [book for book in self.books if title.lower() in book['title'].lower()]
        if found_books:
            print("Books found:")
            for book in found_books:
                status = "Available" if not book["borrowed"] else "Borrowed"
                print(f"Title: {book['title']}, Author: {book['author']}, Year: {book['year']}, Status: {status}")
        else:
            print("No book found with that title.")

    def borrow_book(self, title):
        for book in self.books:
            if book['title'].lower() == title.lower():
                if not book['borrowed']:
                    book['borrowed'] = True
                    print(f"You have successfully borrowed '{book['title']}'.")
                    return
                else:
                    print(f"Sorry, '{book['title']}' is currently borrowed.")
                    return
        print("Book not found.")

    def return_book(self, title):
        for book in self.books:
            if book['title'].lower() == title.lower():
                if book['borrowed']:
                    book['borrowed'] = False
                    print(f"Thank you for returning '{book['title']}'.")
                    return
                else:
                    print(f"'{book['title']}' was not borrowed.")
                    return
        print("Book not found.")

    def view_books(self):
        if not self.books:
            print("No books in the library.")
        else:
            print("Library Books:")
            for book in self.books:
                status = "Available" if not book["borrowed"] else "Borrowed"
                print(f"Title: {book['title']}, Author: {book['author']}, Year: {book['year']}, Status: {status}")


def main():
    library = Library()

    while True:
        print("\nLibrary Management System")
        print("1. Add a new book")
        print("2. Search for a book by title")
        print("3. Borrow a book")
        print("4. Return a book")
        print("5. View all books")
        print("6. Exit")
        choice = input("Enter your choice: ")

        if choice == "1":
            title = input("Enter the book title: ")
            author = input("Enter the book author: ")
            year = input("Enter the year of publication: ")
            library.add_book(title, author, year)

        elif choice == "2":
            title = input("Enter the title of the book to search: ")
            library.search_book(title)

        elif choice == "3":
            title = input("Enter the title of the book to borrow: ")
            library.borrow_book(title)

        elif choice == "4":
            title = input("Enter the title of the book to return: ")
            library.return_book(title)

        elif choice == "5":
            library.view_books()

        elif choice == "6":
            print("Exiting the program. Goodbye!")
            break

        else:
            print("Invalid choice. Please try again.")


if __name__ == "__main__":
    main()
```
